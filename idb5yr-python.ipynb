### 3. Track the rate of natural increase (RNI) over time in the countries

The normal formula to calculate RNI is: (Crude birth rate - Crude death rate) (typically expressed as per 100 population). _(method 1)_

However, the variable descriptions show a different way to calculate RNI, which is: "Difference between births and deaths, divided by the population at midyear and multiplied by 100". _(method 2)_

I will examine how to calculate RNI in two ways, verifying the formula that the United States Census Bureau has written, and see if there is any difference.

````python
rni = idb5yr.loc[:, ['#YR', 'GEO_ID', 'RNI']] # Filter only for necessary columns
rni['RNI_method1'] = (idb5yr['CBR'] - idb5yr['CDR']) / 10 # The crude birth & death rate expressed as per 1000 population
rni['RNI_method2'] = (idb5yr['BIRTHS'] - idb5yr['DEATHS']) / idb5yr['POP'] * 100
rni.head(10)
````

````python
# Countries that had the highest RNI in 2020
rni_2020 = rni[rni['#YR'] == 2020]
rni_2020.sort_values('RNI_method1', ascending = False).head(5)
````

````python
# Countries that had the lowest RNI in 2020
rni_2020 = rni[rni['#YR'] == 2020]
rni_2020.sort_values('RNI_method1', ascending = True).head(5)
````

Comparing the two methods of RNI calculations with the preset RNI values in the data, we can see that the results are slightly different, although these differences are not actually very significant. One possible reason might be related to round-off errors.

### III. Predicting the population of Vietnam based on multiple factors using imputation and regression

````python
# Import the necessary machine learning libraries
from sklearn.experimental import enable_iterative_imputer # this is experimental data, so I need to do this to avoid the warning
from sklearn.impute import IterativeImputer
from sklearn.model_selection import cross_val_score, TimeSeriesSplit
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_squared_error
````

Because this dataset contains variables divided into age groups and sexes, in order to reduce redundancy, I only choose a number of feature variables to predict the population, which are:

- **#YR**: The year in which the population data was recorded.
- **BIRTHS**: Number of live births during the year (for both sexes).
- **CBR**: Crude birth rate = Live births per 1,000 midyear population.
- **CDR**: Crude death rate = Deaths per 1,000 midyear population.
- **DEATHS**: Number of deaths during the year (for both sexes).
- **DEPND**: Dependency ratio = Ratio of population age 0-14 and 65+ to population age 15-64, multiplied by 100.
- **E0**: Life expectancy at birth (for both sexes).
- **GR**: Annual Growth Rate = Change in the population from births, deaths, and net migration, divided by the population at midyear, and multiplied by 100.
- **GRR**: Gross reproduction rate = Average number of daughters females would have by age 49 if the year's age-specific fertility rates and sex-ratio at birth remained constant.
- **IMR**: Infant mortality rate = Number of deaths before reaching age 1 per 1,000 live births (for both sexes).
- **MEDAGE**: Median age in the population (for both sexes).
- **NATINCR**: Natural Increase = The difference between births and deaths.
- **NIM**: Net international migrants = Difference between the number of immigrants entering and emigrants leaving a country in a year (for both sexes).
- **NMR**: Net migration rate = Difference between the number of immigrants entering and emigrants leaving a country in a year, per 1,000 midyear population.
- **POP_DENS**: Population density = Population per square kilometer.
- **RNI**: Rate of natural increase = Difference between births and deaths, divided by the population at midyear and multiplied by 100.
- **SEXRATIO**: Sex ratio of the population = Ratio of males to females in the population at midyear.
- **SRB**: Sex ratio at birth = Males per females born among live births.
- **TFR**: Total fertility rate = Average number of children females would have by age 49 if the year's age-specific fertility rates remained constant.

The target variable to predict is **POP**, the total population recorded at midyear.

For this project, I will test the dataset on a variety of regression models, such as **LinearRegression**, **Ridge**, **Lasso**, and **ElasticNet** in order to choose which best fits the data and predict the population for future years.

My scope of 'existing data' contains Vietnam's recorded demographic metrics between 1950 and 2020, because the data was only updated to 2020.

#### Implementing the model

````python
# Split the dataset into feature and target arrays
X = vn_imputing.drop('POP', axis = 1).values
y = vn_imputing[['POP']].values

# Verify that X and y are NumPy arrays
print(type(X), type(y))
print(X.shape, y.shape)
````

The results of machine learning models will sometimes vary significantly with different choices of training and test sets, so I decided to use a 5-fold **cross-validation** to better evaluate how the models perform in the long run.

The data is already organized in chronological order from 1950-2020, with the ```#YR``` (year) column denoting the time when the data was recorded. For my cross-validation to accurately reflect the real-world implementation of machine learning: training on past data to predict future data, I use the ```TimeSeriesSplit``` method in ```sklearn.model_selection``` to split data into training and test sets. [(link)](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.TimeSeriesSplit.html)

````python
# 5-fold cross-validation for time series
tscv = TimeSeriesSplit(n_splits = 5)
````

````python
# Print the splitting indices that this particular random seed of KFold has done to the dataset
for train_indices, test_indices in tscv.split(X, y):
    print(train_indices)
    print(test_indices)
    print(type(train_indices), type(test_indices))
````

As seen in the result above, ```TimeSeriesSplit``` actually preserves the order of the data, and tests on a fixed number of 'future' observations. As the ```TimesSeriesSplit``` method proceeds to later folds, it expands its training set to more 'recent past' data.

Before testing the population dataset on the regression models, I also use imputation to fill in empty data points and scaling to improve performance.

- ```IterativeImputer()```: Because the missing data is scattered across multiple variables and some variables correlate with each other, I use this imputation method to estimate missing values based on their relationship with other feature variables. _(Note: This is only an experimental estimator, and I need to import another ```sklearn``` library to enable this imputation method.)_
- ```StandardScaler()```: I put all variables on the same scale by subtracting each value in the dataset from the mean and dividing it by the standard deviation of the training samples.

````python
# Set up a function that uses a pipeline to streamline repetitive machine learning workflows and returns the cross-validation scores 
def negative_mse(model_type):
    transformation = [('iterative_imputer', IterativeImputer(random_state = 225275)), ('standard_scaler', StandardScaler())]
    transformation.append(model_type)
    pipeline = Pipeline(transformation)
    return cross_val_score(pipeline, X, y, scoring = 'neg_mean_squared_error')
````

I selected the negative mean squared error (```-MSE```) as the metric to evaluate the performance of different regression models.

MSE penalizes large deviation between the actual values and the values predicted by the model. However, since larger MSE represents a worse model, we can add the negative sign to make our evaluation aligns more with the description of a score, in which larger is better.

````python
# Calculate negative MSE for different models using a function as a syntax shortcut
models = [('linear_regression', LinearRegression()), ('ridge', Ridge()), ('lasso', Lasso()), ('elastic_net', ElasticNet())]
results = []
for model_type in models:
    print(f'Negative MSE for {model_type[0]}: {negative_mse(model_type)}')
    results.append(negative_mse(model_type))
````

````python
# Plot the -MSE values obtained on a graph to visually compare the performance of different models
fig, ax = plt.subplots(figsize = (12, 6))
ax.boxplot(results, tick_labels = [models[0][0], models[1][0], models[2][0], models[3][0]])

plt.title('Negative MSE of different regression models in predicting population data')
plt.xlabel('Regression model')
plt.ylabel('Negative MSE (in 10^12s)')
````

It seems like (multiple) linear regression is the best performing model for predicting population based on the existing demographic features that I have chosen, because it consistently outperforms other regression models in the list, and the list of negative MSE values obtained with time-series cross-validation has a very low variance.

I have trained my model on the demographic information from 1950-2020 (71 years). To avoid extrapolating too far away in time, I will let the linear regression model predict for the next 15 years, a little more than 20% of the dataset (```71 years * (1/5) = 14.2 years```).

#### Use linear regression model to predict the population for the next 15 years

All the demographic metrics such as births, deaths, density, or median age from 2021 onward are actually the projected values. If they continue the same trend that the US Census Bureau has predicted, how much of an increase in the population will Vietnam undergo?

````python
# Extract 'future' data (records for the years 2021-2035)
vn_future_temp = idb5yr_imputing[np.logical_and(idb5yr_imputing['GEO_ID'] == 'W140000WOVN',
                                           np.logical_and(idb5yr_imputing['#YR'] >= 2021, idb5yr_imputing['#YR'] <= 2035))]

vn_future = vn_future_temp[['#YR', 'BIRTHS', 'CBR', 'CDR', 'DEATHS', 'DEPND', 'E0', 'GR', 'GRR', 'IMR', 'MEDAGE', 'NATINCR', 'NIM', 'NMR', 'POP_DENS', 'RNI', 'SEXRATIO', 'SRB', 'TFR', 'POP']]
vn_future
````

````python
# Set up the training and test sets
X = vn_imputing.drop('POP', axis = 1).values
y = vn_imputing[['POP']].values

# I want to compare between the regression models and the actual projected values provided in the dataset by the US Census Bureau
X_future = vn_future.drop('POP', axis = 1).values # values of the features we are using to predict the population (2021-2035)
y_future = vn_future[['POP']].values # values actually predicted in the original data
````

````python
# Show the R^2 for each fold created with TimeSeriesSplit
linreg_transform = Pipeline([('iterative_imputer', IterativeImputer(random_state = 225275, initial_strategy = 'constant')),
                             ('standard_scaler', StandardScaler()),
                             ('linear_regression', LinearRegression())])

for train_indices, test_indices in tscv.split(X, y): # we still use the same time-series split as in the previous part\
    linreg_transform.fit(X[train_indices], y[train_indices])
    print(linreg_transform.score(X[test_indices], y[test_indices]))
````

The regression results will vary slightly for each fold of ```TimeSeriesSplit``` because the training data is slightly different for each fold. I choose to do the prediction on the last fold, which contains more useful information while not having such a low R<sup>2</sup>.

````python
# Show the regression coefficients for the last fold of TimeSeriesSplit
for num, (train_indices, test_indices) in enumerate(tscv.split(X, y)):
    if num <= 3: continue
    else:
        linreg_transform.fit(X[train_indices], y[train_indices])
        coefficients = linreg_transform.named_steps['linear_regression'].coef_
        print(f'Array of regression coefficients: {coefficients}')
        predictions = linreg_transform.predict(X_future)
````

````python
# Show the contribution of each variable to population
fig, ax = plt.subplots(figsize = (12, 6))
ax.bar(vn_future.drop('POP', axis = 1).columns, coefficients.flatten()) # flatten() to make the coefficients array become 1-dimensional
plt.xticks(rotation = 45)

plt.title('Coefficient of each variable in the linear regression model')
plt.xlabel('Variable')
plt.ylabel('Coefficient (in 10^14s)')
plt.axhline(y = 0, linestyle = '--', color = 'red')
````

It seems that year (```#YR```), number of deaths (```DEATHS```), and the Natural Increase (```NATINCR```) of the population are what contributes the most to the change in population, according to linear regression model. As the number of births increase, population is likely to increase. Likewise, as the number of deaths increase, population is likely to decrease.

However, the coefficients of other variables are so small in comparison that we cannot see many bars from this graph. I'd like to draw the graph again to see how **_other_** variables contribute to the regression model.

````python
fig, ax = plt.subplots(figsize = (12, 6))
ax.bar(vn_future.drop(['BIRTHS', 'DEATHS', 'NATINCR', 'POP'], axis = 1).columns,
       np.delete(coefficients.flatten(), [1, 4, 11]))
plt.xticks(rotation = 45)

plt.title('Coefficient of each variable in the linear regression model (zoomed in)')
plt.xlabel('Variable')
plt.ylabel('Coefficient (in 10^7s)')
plt.axhline(y = 0, linestyle = '--', color = 'red')
````

After I zoomed in the graph to see other variables, the increase in population density (```POP_DENS```) and Rate of Natural Increase (```RNI```) also have a considerable increasing impact on the future population. On the other hand, the crude birth rate (```CBR```) has a strong negative impact on the population: an increase in live births per 1,000 midyear population is actually associated with a population decrease.

Sex ratio in the population (```SEXRATIO```) and sex ratio at birth (```SRB```) contributed the least to the regression equation compared to all other variables; we can barely see the bars even after zooming in.

````python
# Plot the values predicted by Linear Regression and the existing projected values in the dataset
fig, ax = plt.subplots(figsize = (12, 6))
plt.plot(range(2021, 2036, 1), predictions, marker = 'o', linestyle = '--') # the model's prediction
plt.plot(range(2021, 2036, 1), y_future, marker = 'o', linestyle = '-') # the prediction in the dataset
plt.legend(['Prediction from multiple linear regression model', 'Existing projected values in the dataset'])
plt.title('Estimated population of Vietnam (2021-2035)')
plt.xlabel('Year')
plt.ylabel('Population (in hundreds of millions)')
````

My multiple linear regression model with more variables actually yield quite a similar result compared to the simple linear regression prediction with the IDB Single Year dataset. The model's prediction aligns closely with the US Census Bureau's projected values at first, but linear regression predicts a larger increase in the population than the existing projections in the dataset does.
