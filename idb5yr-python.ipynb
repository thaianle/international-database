## Analyzing the IDB 5 Year database

### 1. Track the rate of natural increase (RNI) over time in the countries

The normal formula to calculate RNI is: (Crude birth rate - Crude death rate) (typically expressed as per 100 population). _(method 1)_

However, the variable descriptions show a different way to calculate RNI, which is: "Difference between births and deaths, divided by the population at midyear and multiplied by 100". _(method 2)_

I will examine how to calculate RNI in two ways, verifying the formula that the United States Census Bureau has written, and see if there is any difference.

```python
rni = idb5yr.loc[:, ['#YR', 'GEO_ID', 'RNI']] # Filter only for necessary columns
rni['RNI_method1'] = (idb5yr['CBR'] - idb5yr['CDR']) / 10 # The crude birth & death rate expressed as per 1000 population
rni['RNI_method2'] = (idb5yr['BIRTHS'] - idb5yr['DEATHS']) / idb5yr['POP'] * 100
rni.head(10)
```

```python
# Filtering only for "not null" rows
rni = rni[rni['RNI'].notnull()]
rni.head(10)
```

```python
# Countries that had the highest RNI in 2020
rni_2020 = rni[rni['#YR'] == 2020]
rni_2020.sort_values('RNI_method1', ascending = False).head(5)
```

```python
# Countries that had the lowest RNI in 2020
rni_2020 = rni[rni['#YR'] == 2020]
rni_2020.sort_values('RNI_method1', ascending = True).head(5)
```

Comparing the two methods of RNI calculations with the preset RNI values in the data, we can see that the results are slightly different, although these differences are not actually very significant. One possible reason might be related to round-off errors.

### 2. Comparing age distributions at two different points in time

How had the age distribution of Vietnam changed from 1989 to 2020?

```python
vn_comparison = idb5yr[np.logical_and(idb5yr['#YR'].isin([1989, 2020]), idb5yr['GEO_ID'] == 'W140000WOVN')]
vn_comparison
```

```python
# Reshape the table from wide to long format for easier plotting and comparison
vn_comparison_long = vn_comparison.melt(id_vars = ['#YR', 'GEO_ID'])
vn_comparison_long.head(10)
```

```python
# Choose the necessary variables only
vn_1989 = vn_comparison_long[vn_comparison_long['#YR'] == 1989]
vn_1989 = vn_1989[vn_1989['variable'].isin(['FPOP0_4', 'FPOP5_9', 'FPOP10_14', 'FPOP15_19', 'FPOP20_24', 'FPOP25_29', 'FPOP30_34', 'FPOP35_39',
                                          'FPOP40_44', 'FPOP45_49', 'FPOP50_54', 'FPOP55_59', 'FPOP60_64', 'FPOP65_69', 'FPOP70_74', 'FPOP75_79',
                                          'FPOP80_84', 'FPOP85_89', 'FPOP90_94', 'FPOP95_99', 'FPOP100_'])]
vn_1989.head(10)
```

```python
# Set index and rearrange the indexes to have the age groups in order
vn_1989_arr = vn_1989.set_index('variable')
vn_1989_arr = vn_1989_arr.reindex(['FPOP0_4', 'FPOP5_9', 'FPOP10_14', 'FPOP15_19', 'FPOP20_24', 'FPOP25_29', 'FPOP30_34', 'FPOP35_39',
                     'FPOP40_44', 'FPOP45_49', 'FPOP50_54', 'FPOP55_59', 'FPOP60_64', 'FPOP65_69', 'FPOP70_74', 'FPOP75_79',
                     'FPOP80_84', 'FPOP85_89', 'FPOP90_94', 'FPOP95_99', 'FPOP100_'])

# Divide each value by the total to have the percentage
vn_1989_arr['pct'] = vn_1989_arr['value'] / vn_1989_arr['value'].sum()
vn_1989_arr.head(10)
```

```python
vn_1989_plot = sns.catplot(data = vn_1989_arr, x = 'variable', y = 'pct', kind = 'bar', aspect = 2)
vn_1989_plot.set_xticklabels(['0-4', '5-9', '10-14', '15-19', '20-24', '25-29', '30-34', '35-39', '40-44', '45-49', '50-54', '55-59',
                              '60-64', '65-69', '70-74', '75-79', '80-84', '85-89', '90-94', '95-99', '100+'])

vn_1989_plot.set_axis_labels('Age group', 'Percentage')
vn_1989_plot.fig.suptitle('Female age distribution of Vietnam in 1989', y = 1.05)
```

The age distribution of Vietnam in 1989 is heavily skewed right when I arranged the age groups in order. Overall, the country had a young population at the time, with over 30% of the population below 20 years old.

As we progress to older age groups, their proportion in the population decreased steadily, with the steepest decline happening from the 25-29 to 40-44 year old group.

```python
# Do the same with 2020 data
vn_2020 = vn_comparison_long[vn_comparison_long['#YR'] == 2020]
vn_2020 = vn_2020[vn_2020['variable'].isin(['FPOP0_4', 'FPOP5_9', 'FPOP10_14', 'FPOP15_19', 'FPOP20_24', 'FPOP25_29', 'FPOP30_34', 'FPOP35_39',
                                          'FPOP40_44', 'FPOP45_49', 'FPOP50_54', 'FPOP55_59', 'FPOP60_64', 'FPOP65_69', 'FPOP70_74', 'FPOP75_79',
                                          'FPOP80_84', 'FPOP85_89', 'FPOP90_94', 'FPOP95_99', 'FPOP100_'])]

vn_2020_arr = vn_2020.set_index('variable')
vn_2020_arr = vn_2020_arr.reindex(['FPOP0_4', 'FPOP5_9', 'FPOP10_14', 'FPOP15_19', 'FPOP20_24', 'FPOP25_29', 'FPOP30_34', 'FPOP35_39',
                     'FPOP40_44', 'FPOP45_49', 'FPOP50_54', 'FPOP55_59', 'FPOP60_64', 'FPOP65_69', 'FPOP70_74', 'FPOP75_79',
                     'FPOP80_84', 'FPOP85_89', 'FPOP90_94', 'FPOP95_99', 'FPOP100_'])

vn_2020_arr['pct'] = vn_2020_arr['value'] / vn_2020_arr['value'].sum()
vn_2020_arr.head(10)
```

```python
vn_2020_plot = sns.catplot(data = vn_2020_arr, x = 'variable', y = 'pct', kind = 'bar', aspect = 2)
vn_2020_plot.set_xticklabels(['0-4', '5-9', '10-14', '15-19', '20-24', '25-29', '30-34', '35-39', '40-44', '45-49', '50-54', '55-59',
                              '60-64', '65-69', '70-74', '75-79', '80-84', '85-89', '90-94', '95-99', '100+'])

vn_2020_plot.set_axis_labels('Age group', 'Percentage')
vn_2020_plot.fig.suptitle('Female age distribution of Vietnam in 2020', y = 1.05)
plt.ylim(0.00, 0.14) # Adjust the axis to fit the graph for 1989
```

Compared to 1989, Vietnam had transitioned to an older age distribution as of 2020. The middle-aged groups (40 to 64 years old) took up a much larger share of the country's population. The opposite trend was seen in the young-aged groups (0 to 19 years old) in which their percentage in the female population declined.

I am trying to merge these two graphs together, and the work is still in progress.

```python
vn_fpop = vn_comparison_long[vn_comparison_long['variable'].isin(['FPOP0_4', 'FPOP5_9', 'FPOP10_14', 'FPOP15_19', 'FPOP20_24', 'FPOP25_29', 'FPOP30_34', 'FPOP35_39',
                                          'FPOP40_44', 'FPOP45_49', 'FPOP50_54', 'FPOP55_59', 'FPOP60_64', 'FPOP65_69', 'FPOP70_74', 'FPOP75_79',
                                          'FPOP80_84', 'FPOP85_89', 'FPOP90_94', 'FPOP95_99', 'FPOP100_'])]
vn_fpop.head(10)
```

```python
# Divide each value by the total to have the percentage
vn_fpop_arr['pct_1989'] = vn_fpop_arr['value'] / vn_fpop_arr[vn_fpop_arr['#YR'] == 1989]['value'].sum()
vn_fpop_arr['pct_2020'] = vn_fpop_arr['value'] / vn_fpop_arr[vn_fpop_arr['#YR'] == 2020]['value'].sum()

# Set null for the percentages in the "wrong year"
vn_fpop_arr.loc[vn_fpop_arr['#YR'] == 1989, 'pct_2020'] = np.nan
vn_fpop_arr.loc[vn_fpop_arr['#YR'] == 2020, 'pct_1989'] = np.nan
vn_fpop_arr['pct_merged'] = vn_fpop_arr['pct_1989'].combine_first(vn_fpop_arr['pct_2020'])

vn_fpop_arr
```

```python
vn_plot = sns.catplot(data = vn_fpop_arr, x = 'variable', y = 'pct_merged', hue = '#YR', kind = 'bar', aspect = 2)
vn_plot.set_xticklabels(['0-4', '5-9', '10-14', '15-19', '20-24', '25-29', '30-34', '35-39', '40-44', '45-49', '50-54', '55-59',
                              '60-64', '65-69', '70-74', '75-79', '80-84', '85-89', '90-94', '95-99', '100+'])

vn_1989_plot.set_axis_labels('Age group', 'Percentage')
vn_1989_plot.fig.suptitle('Female age distribution of Vietnam in 1989', y = 1.05)
```
