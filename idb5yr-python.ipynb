## II. Analyzing the IDB 5 Year database

### 1. Track the rate of natural increase (RNI) over time in the countries

The normal formula to calculate RNI is: (Crude birth rate - Crude death rate) (typically expressed as per 100 population). _(method 1)_

However, the variable descriptions show a different way to calculate RNI, which is: "Difference between births and deaths, divided by the population at midyear and multiplied by 100". _(method 2)_

I will examine how to calculate RNI in two ways, verifying the formula that the United States Census Bureau has written, and see if there is any difference.

```python
rni = idb5yr.loc[:, ['#YR', 'GEO_ID', 'RNI']] # Filter only for necessary columns
rni['RNI_method1'] = (idb5yr['CBR'] - idb5yr['CDR']) / 10 # The crude birth & death rate expressed as per 1000 population
rni['RNI_method2'] = (idb5yr['BIRTHS'] - idb5yr['DEATHS']) / idb5yr['POP'] * 100
rni.head(10)
```

```python
# Countries that had the highest RNI in 2020
rni_2020 = rni[rni['#YR'] == 2020]
rni_2020.sort_values('RNI_method1', ascending = False).head(5)
```

```python
# Countries that had the lowest RNI in 2020
rni_2020 = rni[rni['#YR'] == 2020]
rni_2020.sort_values('RNI_method1', ascending = True).head(5)
```

Comparing the two methods of RNI calculations with the preset RNI values in the data, we can see that the results are slightly different, although these differences are not actually very significant. One possible reason might be related to round-off errors.

### 2. How had the age structure of Vietnam changed from 1990 to 2020?

In this table, the male and female populations measured at midyear are divided into 5-year segments (hence the "IDB 5 Years" name). The age segments are measured by ranges of integers: 0-4 years, 5-9 years, 10-14 years, ..., 95-99 years, and 100+ years.

The column names of interest are:

- **FPOP0_4**: Female midyear population age 0-4 years
- **FPOP5_9**: Female midyear population age 5-9 years
- **FPOP10_14**: Female midyear population age 10-14 years
- ...
- **FPOP95_99**: Female midyear population age 95-99 years
- **FPOP100_**: Female midyear population age 100+ years

*(From the original variable descriptions by the Census Bureau)*

The column names for male midyear population follow the similar naming convention, but they start with **'MPOP'** instead of **'FPOP'**.

```python
vn_comparison = idb5yr[np.logical_and(idb5yr['#YR'].isin([1990, 2020]), idb5yr['GEO_ID'] == 'W140000WOVN')]
vn_comparison
```

```python
# Reshape the table from wide to long format for easier plotting and comparison
vn_comparison_long = vn_comparison.melt(id_vars = ['#YR', 'GEO_ID'])
vn_comparison_long.head(10)
```

```python
vn_fpop = vn_comparison_long[vn_comparison_long['variable'].isin(['FPOP0_4', 'FPOP5_9', 'FPOP10_14', 'FPOP15_19', 'FPOP20_24', 'FPOP25_29', 'FPOP30_34', 'FPOP35_39',
                                          'FPOP40_44', 'FPOP45_49', 'FPOP50_54', 'FPOP55_59', 'FPOP60_64', 'FPOP65_69', 'FPOP70_74', 'FPOP75_79',
                                          'FPOP80_84', 'FPOP85_89', 'FPOP90_94', 'FPOP95_99', 'FPOP100_'])]
vn_fpop.head(10)
```

```python
# Copy the DataFrame to avoid SettingWithCopyWarning
vn_fpop_reshaped = vn_fpop.copy()

# Divide each value by the total to have the percentage
vn_fpop_reshaped['pct_1990'] = vn_fpop_reshaped['value'] / vn_fpop_reshaped[vn_fpop_reshaped['#YR'] == 1990]['value'].sum()
vn_fpop_reshaped['pct_2020'] = vn_fpop_reshaped['value'] / vn_fpop_reshaped[vn_fpop_reshaped['#YR'] == 2020]['value'].sum()

# Set null for the percentages in the "wrong year"
vn_fpop_reshaped.loc[vn_fpop_reshaped['#YR'] == 1990, 'pct_2020'] = np.nan
vn_fpop_reshaped.loc[vn_fpop_reshaped['#YR'] == 2020, 'pct_1990'] = np.nan
vn_fpop_reshaped['pct_merged'] = vn_fpop_reshaped['pct_1990'].combine_first(vn_fpop_reshaped['pct_2020'])

# Rearranging age groups in an arbitrary order to fit the age distribution
age_order = ['FPOP0_4', 'FPOP5_9', 'FPOP10_14', 'FPOP15_19', 'FPOP20_24', 'FPOP25_29', 'FPOP30_34', 'FPOP35_39',
             'FPOP40_44', 'FPOP45_49', 'FPOP50_54', 'FPOP55_59', 'FPOP60_64', 'FPOP65_69', 'FPOP70_74', 'FPOP75_79',
             'FPOP80_84', 'FPOP85_89', 'FPOP90_94', 'FPOP95_99', 'FPOP100_']

vn_fpop_reshaped['variable'].astype('category')
vn_fpop_reshaped['variable'] = pd.Categorical(vn_fpop_reshaped['variable'], categories = age_order, ordered = True)
vn_fpop_reshaped = vn_fpop_reshaped.sort_values(['variable', '#YR'])

vn_fpop_reshaped.head(10)
```

```python
vn_plot_f = sns.catplot(data = vn_fpop_reshaped, x = 'variable', y = 'pct_merged', hue = '#YR', kind = 'bar', aspect = 2, palette = 'flare')
vn_plot_f.set_xticklabels(['0-4', '5-9', '10-14', '15-19', '20-24', '25-29', '30-34', '35-39', '40-44', '45-49', '50-54', '55-59',
                              '60-64', '65-69', '70-74', '75-79', '80-84', '85-89', '90-94', '95-99', '100+'])

plt.ylim(0, 0.15)
plt.title('Female population distribution by age in Vietnam', y = 1.05)
plt.xlabel('Age group')
plt.ylabel('Proportion')
```

The age distribution of female populations in Vietnam in 1990 is heavily skewed right when I arranged the age groups in order. Overall, the country had a young population at the time, with over 30% of the population below 20 years old.

As we progress to older age groups, their proportion in the population decreased steadily, with the steepest decline happening from the 25-29 to 40-44 year old group.

Compared to 1990, Vietnam had transitioned to an older age distribution as of 2020. The middle-aged groups (40 to 64 years old) took up a much larger share of the country's population. The opposite trend was seen in the young-aged groups (0 to 19 years old) in which their percentage in the female population declined.

### 3. Does sex ratio at birth impact sex ratio in the population of Vietnam?

Column names of interest:
- **SEXRATIO**: Sex ratio of the population: Ratio of males to females in the population at midyear
- **SRB**: Sex ratio at birth: Males per females born among live births

*(From the original variable descriptions by the Census Bureau)*

```python
# Filter for sex ratio at birth and for sex ratio in the population
vn = idb5yr[idb5yr['GEO_ID'] == 'W140000WOVN']
vn = vn.melt(id_vars = ['#YR', 'GEO_ID']) # change from wide to long data
srvn = vn[vn['variable'].isin(['SRB', 'SEXRATIO'])] # filter only the necessary variables
srvn
```

```python
srvn_plot = sns.relplot(data = srvn, x = '#YR', y = 'value', kind = 'line', hue = 'variable', aspect = 2, marker = 'o', legend = None)
plt.xlabel('Year')
plt.ylabel('Sex ratio (Male / Female)')

# Add a horizontal line to denote equality (sex ratio: 1 male / 1 female)
# Add a vertical line to denote the projections
plt.axvline(x = 2020, linestyle = '-.', color = 'gray')
plt.axhline(y = 1, linestyle = '-', color = 'black')

# Customize title and legends
plt.title('Sex ratio at birth and sex ratio in the population of Vietnam (1989-projected 2100)', y = 1.05)
plt.legend(['Sex ratio at birth', 'Sex ratio in the population'])
```

The sex ratio in the population (the orange curve) consistently increased from 1989 to 2020. From 2018 onward, Vietnam had more male than female in the population. Based on the projections, the sex ratio in the population will continue to increase up until 2030 where the curve flattens out.

Regarding the sex ratio at birth (the blue curve), the ratio was consistently a little more than 1.05. It is unlikely that the sex ratio at birth would not change from year to year, so I think the sex ratio at birth might have been mainly imputed due to insufficient data. From the late 2000s to 2020, the sex ratio at birth fluctuated a lot, however, it underwent an overall increasing trend. The sex ratio at birth peaked in 2019 with approximately 1.15 male / 1 female.
